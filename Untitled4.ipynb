```python
import gradio as gr
import whisper

# Load English speech recognition model
model = whisper.load_model("small")

def transcribe_and_analyze(audio):
    if audio is None:
        return "No recording received.", ""

    # Speech → Text
    result = model.transcribe(
        audio,
        language="en",
        fp16=False
    )

    text = result["text"].strip()

    if not text:
        return "No speech detected.", ""

    # Prototype AI analysis
    analysis = f"""
JANBAK AI — SECURITY ANALYSIS

Report:
{text}

Incident Type:
Suspicious Activity

Risk Level:
HIGH

Key Information:
• Voice report received
• Report successfully transcribed
• Incident information extracted

Recommended Action:
Forward the report to the relevant authority
for verification and appropriate action.
"""

    return text, analysis


with gr.Blocks(
    title="JANBAK AI"
) as demo:

    gr.Markdown(
        "# JANBAK AI\n"
        "### AI-Powered Security Reporting System"
    )

    gr.Markdown(
        "## 📞 Incoming Report\n"
        "The reporter communicates with JANBAK AI by voice."
    )

    audio = gr.Audio(
        sources=["microphone"],
        type="filepath",
        label="🎙️ Reporter Voice"
    )

    button = gr.Button(
        "Analyze Report",
        variant="primary"
    )

    transcription = gr.Textbox(
        label="📝 Report Transcription",
        lines=5
    )

    analysis = gr.Textbox(
        label="🤖 JANBAK AI Analysis",
        lines=12
    )

    button.click(
        fn=transcribe_and_analyze,
        inputs=audio,
        outputs=[transcription, analysis]
    )

demo.launch(share=True)
```

```python
def analyze_report_text(text):
    text_lower = text.lower()

    if "fire" in text_lower:
        incident = "Fire / Possible Fire Emergency"
        risk = "CRITICAL"
        authority = "Emergency Response Authority"
        action = "Immediate emergency response recommended."
    else:
        incident = "Security Incident"
        risk = "MEDIUM"
        authority = "Relevant Security Authority"
        action = "Report should be reviewed by the relevant authority."

    return f"""
JANBAK AI — INCIDENT ANALYSIS

Incident:
{incident}

Risk Level:
{risk}

Detected Information:
{text}

Assigned Authority:
{authority}

Recommended Action:
{action}
"""


with gr.Blocks(title="JANBAK AI Analysis") as demo:

    gr.Markdown("# JANBAK AI")
    gr.Markdown("### Automated Security Incident Analysis")

    report = gr.Textbox(
        label="Reporter Message",
        placeholder="Enter the transcribed report..."
    )

    analyze = gr.Button(
        "Analyze Incident",
        variant="primary"
    )

    result = gr.Textbox(
        label="AI Analysis",
        lines=12
    )

    analyze.click(
        analyze_report_text,
        inputs=report,
        outputs=result
    )

demo.launch(share=True)
```

```python
import gradio as gr
import whisper
import re

# Speech recognition model
model = whisper.load_model("small")


def process_report(audio, phone):

    if audio is None:
        return (
            "No voice report received.",
            "",
            "",
            "",
            ""
        )

    # 1 — Speech to text
    result = model.transcribe(
        audio,
        language="en",
        fp16=False
    )

    text = result["text"].strip()

    if not text:
        return (
            "No speech detected.",
            "",
            "",
            "",
            ""
        )

    # 2 — Prototype incident analysis
    text_lower = text.lower()

    if any(word in text_lower for word in ["fire", "burning", "smoke"]):
        incident = "Fire / Emergency"
        risk = "CRITICAL"
        authority = "Emergency Response Authority"
        action = "Immediate emergency response recommended."

    elif any(word in text_lower for word in ["attack", "weapon", "threat"]):
        incident = "Security Threat"
        risk = "HIGH"
        authority = "Security Authority"
        action = "Immediate verification and security response recommended."

    elif any(word in text_lower for word in ["suspicious", "strange"]):
        incident = "Suspicious Activity"
        risk = "MEDIUM"
        authority = "Security Authority"
        action = "Verify the reported activity."

    else:
        incident = "General Security Report"
        risk = "MEDIUM"
        authority = "Relevant Authority"
        action = "Review and verify the report."

    # 3 — Authority report
    report_id = "JAN-001"

    authority_report = f"""
JANBAK AI — AUTHORITY REPORT

Report ID:
{report_id}

Reporter:
{phone if phone else "Anonymous"}

Incident:
{incident}

Risk Level:
{risk}

Detected Information:
{text}

Assigned Authority:
{authority}

Recommended Action:
{action}

Status:
REQUIRES REVIEW
"""

    return text, incident, risk, authority, authority_report


with gr.Blocks(
    title="JANBAK AI",
    theme=gr.themes.Soft()
) as demo:

    gr.Markdown("""
    # JANBAK AI
    ### AI-Powered Security Reporting System
    """)

    gr.Markdown("""
    **Reporter → JANBAK AI → Analysis → Relevant Authority**
    """)

    # =========================
    # REPORTER
    # =========================

    gr.Markdown("## 📞 1. Incoming Report")

    phone = gr.Textbox(
        label="Reporter Phone Number",
        placeholder="+966 5X XXX XXXX"
    )

    audio = gr.Audio(
        sources=["microphone"],
        type="filepath",
        label="🎙️ Voice Report"
    )

    analyze_button = gr.Button(
        "🚨 Send & Analyze Report",
        variant="primary"
    )

    # =========================
    # AI ANALYSIS
    # =========================

    gr.Markdown("## 🤖 2. JANBAK AI Analysis")

    transcription = gr.Textbox(
        label="Speech Transcription",
        lines=4
    )

    incident = gr.Textbox(
        label="Incident Type"
    )

    risk = gr.Textbox(
        label="Risk Level"
    )

    authority = gr.Textbox(
        label="Assigned Authority"
    )

    # =========================
    # AUTHORITY
    # =========================

    gr.Markdown("## 🏢 3. Authority Dashboard")

    authority_report = gr.Textbox(
        label="📋 Generated Authority Report",
        lines=16
    )

    analyze_button.click(
        fn=process_report,
        inputs=[audio, phone],
        outputs=[
            transcription,
            incident,
            risk,
            authority,
            authority_report
        ]
    )

demo.launch(share=True)
```
